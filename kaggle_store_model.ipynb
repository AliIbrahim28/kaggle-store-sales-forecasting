```python
# ============================================
# Merge transactions into TRAIN only (for EDA/comparison purposes)
# NOT merged into test — it doesn't exist for future dates
# ============================================

train = train.merge(transactions, on=['date', 'store_nbr'], how='left')

print("Missing transactions in train:", train['transactions'].isna().sum())
print("Missing transactions %:", train['transactions'].isna().mean())

# Fill missing transactions (only within train) using median
# Median is safer than mean here since transaction counts can have outlier spikes
train['transactions'] = train['transactions'].fillna(train['transactions'].median())

print("After fill, missing transactions in train:", train['transactions'].isna().sum())
```

```python


X = train.drop(columns=['id', 'sales', 'date'])
y = train['sales']

print(X.shape)
print(y.shape)
print(X.columns.tolist())
```

```python
dates = train['date']   # keep this handy for the split step
```

```python
cutoff_date = dates.max() - pd.Timedelta(days=60)

train_mask = dates <= cutoff_date
test_mask = dates > cutoff_date

X_train, y_train = X[train_mask], y[train_mask]
X_test, y_test = X[test_mask], y[test_mask]

print("X_train:", X_train.shape, "y_train:", y_train.shape)
print("X_test:", X_test.shape, "y_test:", y_test.shape)
```

```python


categorical_cols = ['family', 'city', 'state', 'type']
numeric_cols = ['store_nbr', 'onpromotion', 'dcoilwtico', 'is_holiday', 'cluster',
                 'year', 'month', 'day', 'day_of_week', 'is_weekend',
                 'week_of_year', 'quarter', 'is_month_start', 'is_month_end']

# Rebuild X and y fresh (using the original train, not the one-hot encoded version)
X = train[numeric_cols + categorical_cols].copy()
y = train['sales']
dates = train['date']

# Convert categorical columns to pandas 'category' dtype — LightGBM handles this natively
for col in categorical_cols:
    X[col] = X[col].astype('category')

print(X.shape)
print(X.dtypes)
```

```python
cutoff_date = dates.max() - pd.Timedelta(days=60)

train_mask = dates <= cutoff_date
test_mask = dates > cutoff_date

X_train, y_train = X[train_mask], y[train_mask]
X_test, y_test = X[test_mask], y[test_mask]

print("X_train:", X_train.shape, "y_train:", y_train.shape)
print("X_test:", X_test.shape, "y_test:", y_test.shape)
```

```python
import numpy as np

y_train_log = np.log1p(y_train)   # log(1 + sales) — handles sales=0 safely
```

```python
import lightgbm as lgb
from sklearn.metrics import mean_squared_log_error
import time

start = time.time()

model = lgb.LGBMRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=-1,        # let it grow naturally, no artificial cap this time
    num_leaves=63,
    random_state=42
)

model.fit(
    X_train, y_train_log,
    categorical_feature=categorical_cols
)

print(f"Training took {time.time() - start:.1f} seconds")
```

```python
y_pred_log = model.predict(X_test)
y_pred = np.expm1(y_pred_log)          # reverse log1p
y_pred = np.clip(y_pred, 0, None)      # sales can't be negative

rmsle = np.sqrt(mean_squared_log_error(y_test, y_pred))
print("Test RMSLE:", rmsle)

# Also check train performance
y_train_pred = np.clip(np.expm1(model.predict(X_train)), 0, None)
train_rmsle = np.sqrt(mean_squared_log_error(y_train, y_train_pred))
print("Train RMSLE:", train_rmsle)
```

```python
# Combine train_split + test_split back into the full set
X_full = X  # the complete X we built before splitting
y_full_log = np.log1p(y)  # log-transform the full target

model_final = lgb.LGBMRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=-1,
    num_leaves=63,
    random_state=42
)

model_final.fit(X_full, y_full_log, categorical_feature=categorical_cols)
```

```python
X_submit = test[numeric_cols + categorical_cols].copy()

for col in categorical_cols:
    X_submit[col] = X_submit[col].astype('category')

print(X_submit.shape)
print(X_submit.dtypes)
```

```python
y_submit_log = model_final.predict(X_submit)
y_submit = np.expm1(y_submit_log)       # reverse log1p
y_submit = np.clip(y_submit, 0, None)   # sales can't be negative
```

```python
submission = pd.DataFrame({
    'id': test['id'],
    'sales': y_submit
})

print(submission.head())
print(submission.shape)   # should match test.csv's row count exactly
```
